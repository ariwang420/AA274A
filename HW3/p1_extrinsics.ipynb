## Problem 1.2

```python
# 1)  Compute the homography, H.
#     1.a) Use the SQUARE_SIZE, and the dimensions of the grid above to
#             to create a the array P = [[X_0, Y_0, 1], .... [X_n, Y_n, 1]].
#             P should have shape (63, 3).
#          Hint: Use np.meshgrid to generate the X, Y coordinates, and
#             array.reshape and np.hstack to form the final matrix.
#     1.b) Form the M matrix outlined in the write-up!
#     1.c) Use np.linalg.svd(M) to solve for the nullspace of M and recover H.
#          Hint: np.linalg.svd(M) returns V.T.
#          Hint: Use <array>.reshape(3, 3) to give H the proper dimensions.
# 2)  Compute the camera pose rotation, R, and translation, t.
#     2.a) Compute KinvH = inv(K) @ H.
#          Hint: Use np.linalg.inv().
#     2.b) Normalize the values of KinvH to it's first column.
#          Hint: Use np.linalg.norm() and the / operator.
#     2.c) Compute r0, r1, r2, and t as outlined in the write-up.
#          Hint: Use np.cross().
#     2.d) Form R from r0, r1, and r2.
#          Hint: Use np.column_stack()

# --------- YOUR CODE STARTS HERE ---------------
P = np.meshgrid(np.arange(ncorners_x)*SQUARE_SIZE, np.arange(ncorners_y)*SQUARE_SIZE) # shape (63,3), each is x,y,z where z is 0
P = np.column_stack((P[0].reshape(-1), P[1].reshape(-1), np.ones(ncorners_x*ncorners_y)))
N = 63 # number of corners
corners = corners.reshape(N, 2) # shape (63,2) each is u,v

X = P[:, 0] # extract x coord in world space
Y = P[:, 1] # extract x coord in world space
u = corners[:, 0] # extract u from images
v = corners[:, 1] # extract v from images

M = np.zeros((2 * N, 9), dtype=float) # M is the stacked 2x9 matrix for all 63 data pointsF

# each datapoint makes two rows
rows0 = 2 * np.arange(N)
rows1 = rows0 + 1

# row 0 looks like: [-X -Y -1  0 0 0  uX uY u]
M[rows0, 0] = -X
M[rows0, 1] = -Y
M[rows0, 2] = -1
M[rows0, 6] = u * X
M[rows0, 7] = u * Y
M[rows0, 8] = u

# row 1 looks like: [0 0 0 -X -Y -1 vX vY v]
M[rows1, 3] = -X
M[rows1, 4] = -Y
M[rows1, 5] = -1
M[rows1, 6] = v * X
M[rows1, 7] = v * Y
M[rows1, 8] = v

# Use SVD to solve for H
U, S, Vt = np.linalg.svd(M)

# h is nullspace solution, aka the last row of Vt
h = Vt[-1, :]  # shape (9,)

# H is 3x3 homography matrix
H = h.reshape(3, 3)
H = H / H[2, 2] #normalize

# Compute KinvH = inv(K) @ H
KinvH = np.linalg.inv(K) @ H
# Normalize the values of KinvH to it's first column.
KinvH = KinvH / np.linalg.norm(KinvH[:, 0])

r0, r1 = KinvH[:, 0], KinvH[:, 1] # no need for lambda because we already normalized
r2 = np.cross(r0, r1)

R = np.column_stack([r0, r1, r2])
t = KinvH[:, 2]

# --------- YOUR CODE ENDS HERE -----------------
```
